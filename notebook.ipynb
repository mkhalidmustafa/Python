```python
import pandas as pd

data = {
    "Squad": ["Channels", "Qitaf", "IoT", "Tamayouz"],
    "Planned": [100, 90, 80, 95],
    "Actual": [85, 95, 70, 90]
}

df = pd.DataFrame(data)

display(df)
```

```python
import matplotlib.pyplot as plt

df.plot(
    x="Squad",
    y=["Planned", "Actual"],
    kind="bar",
    figsize=(9, 5)
)

plt.title("Squad Delivery Performance")
plt.ylabel("Progress (%)")
plt.xlabel("Squad")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```
